```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine
```

**Step1: Loading and exploring dataset**

```python
# loading csv file through pandas
wal_df = pd.read_csv('Walmart.csv',encoding_errors= 'ignore')
```

```python
wal_df.shape
```

```python
# finding total duplicates
wal_df.duplicated().sum()
```

```python
# removing duplicated from data itself
wal_df.drop_duplicates(inplace=True)
```

```python
# finding null values
wal_df.isna().sum()
```

```python
# removing null values from data itself
wal_df.dropna(inplace=True)
```

```python
wal_df.columns = wal_df.columns.str.lower()
wal_df.columns
```

```python
# getting total sales
wal_df['total'] = wal_df['unit_price'].str.replace('$', '').astype(float) * wal_df['quantity']
```

```python
# rearranging columns
wal_df = wal_df[['invoice_id',	'Branch',	'City',	'category',	'unit_price',	'quantity',	'total',	'date',	'time',	'payment_method',	'rating',	'profit_margin']]
```

```python
# creating engine to connect dataframe with mysql database
mysql_engine = create_engine("mysql+pymysql://root:Root#342817@127.0.0.1:3306/walmart_db")
```

```python
wal_df['total'] = wal_df['total'].round(2)
```

```python
# creating table and inserting values in the table in the selected database
wal_df.to_sql(name='billing_info', index=False, if_exists='append', con=mysql_engine)
```
